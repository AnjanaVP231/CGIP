```python
#1
import matplotlib.pyplot as plt

x1, y1 = map(int, input("Enter x1, y1: ").split())
x2, y2 = map(int, input("Enter x2, y2: ").split())

dx = x2 - x1
dy = y2 - y1

steps = max(abs(dx), abs(dy))

x_inc = dx / steps
y_inc = dy / steps


X, Y = [], []

for i in range(steps + 1):
    X.append(round(x1))
    Y.append(round(y1))
    x1 += x_inc
    y1 += y_inc

plt.plot(X, Y, 'ro-')
plt.title("DDA Line Drawing")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()


plt.show()
```

```python
#2
import matplotlib.pyplot as plt

x1, y1 = map(int, input("Enter x1, y1: ").split())
x2, y2 = map(int, input("Enter x2, y2: ").split())

dx = abs(x2 - x1)
dy = abs(y2 - y1)

sx = 1 if x1 < x2 else -1
sy = 1 if y1 < y2 else -1

err = dx - dy

X, Y = [], []

while True:
    X.append(x1)
    Y.append(y1)

    if x1 == x2 and y1 == y2:
        break

    e2 = 2 * err

    if e2 > -dy:
        err -= dy
        x1 += sx

    if e2 < dx:
        err += dx
        y1 += sy

plt.plot(X, Y, 'bo-')
plt.title("Bresenham Line Drawing")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.axis("equal")
plt.show()
```

```python
#3
import matplotlib.pyplot as plt

xc, yc = map(int, input("Enter center: ").split())
r = int(input("Enter radius: "))

x = 0
y = r
p = 1 - r

X, Y = [], []

while x <= y:
    points = [
        (xc+x, yc+y), (xc-x, yc+y),
        (xc+x, yc-y), (xc-x, yc-y),
        (xc+y, yc+x), (xc-y, yc+x),
        (xc+y, yc-x), (xc-y, yc-x)
    ]

    for px, py in points:
        X.append(px)
        Y.append(py)

    x += 1

    if p < 0:
        p += 2*x + 1
    else:
        y -= 1
        p += 2*(x-y) + 1

plt.scatter(X, Y)
plt.title("Midpoint Circle Drawing")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.axis("equal")
plt.show()
```

```python
#4
import numpy as np
import matplotlib.pyplot as plt

obj = np.array([
    [0, 0, 1],
    [4, 0, 1],
    [2, 3, 1],
    [0, 0, 1]  ])

tx = float(input("Enter tx: "))
ty = float(input("Enter ty: "))

T = np.array([
    [1, 0, tx],
    [0, 1, ty],
    [0, 0, 1]  ])

new = (T @ obj.T).T

plt.plot(obj[:,0], obj[:,1], 'bo-', label="Original")
plt.plot(new[:,0], new[:,1], 'ro-', label="Translated")
plt.title("2D Translation")
plt.legend()
plt.grid()
plt.axis("equal")
plt.show()
```

```python
import matplotlib.pyplot as plt

x1, y1 = map(int, input("Enter first point: ").split())
x2, y2 = map(int, input("Enter second point: ").split())

xmin, ymin, xmax, ymax = 0, 0, 100, 100

def code(x, y):
    c = 0
    if x < xmin: c |= 1
    if x > xmax: c |= 2
    if y < ymin: c |= 4
    if y > ymax: c |= 8
    return c

c1, c2 = code(x1, y1), code(x2, y2)

while True:
    if not (c1 | c2):
        print("Line accepted")
        break
    elif c1 & c2:
        print("Line rejected")
        break
    else:
        c = c1 if c1 else c2

        if c & 8:
            x, y = x1+(x2-x1)*(ymax-y1)/(y2-y1), ymax
        elif c & 4:
            x, y = x1+(x2-x1)*(ymin-y1)/(y2-y1), ymin
        elif c & 2:
            x, y = xmax, y1+(y2-y1)*(xmax-x1)/(x2-x1)
        else:
            x, y = xmin, y1+(y2-y1)*(xmin-x1)/(x2-x1)

        if c == c1:
            x1, y1 = x, y
            c1 = code(x1, y1)
        else:
            x2, y2 = x, y
            c2 = code(x2, y2)

plt.plot([0,100,100,0,0], [0,0,100,100,0], 'k-')

if not (c1 | c2):
    plt.plot([x1,x2], [y1,y2], 'ro-')

plt.title("Cohen-Sutherland Clipping")
plt.grid()
plt.show()
```

```python
#12
import cv2
import numpy as np
import matplotlib.pyplot as plt

path = input("Enter image path: ")

img = cv2.imread(path, 0)

if img is None:
    raise FileNotFoundError("Image not found")

# Gradient kernels
gx = np.array([[-1, 0, 1],
               [-1, 0, 1],
               [-1, 0, 1]])

gy = np.array([[-1, -1, -1],
               [ 0,  0,  0],
               [ 1,  1,  1]])

Gx = cv2.filter2D(img, cv2.CV_32F, gx)
Gy = cv2.filter2D(img, cv2.CV_32F, gy)

gradient = cv2.magnitude(Gx, Gy)

# Laplacian kernels
k1 = np.array([[0, 1, 0],
               [1, -4, 1],
               [0, 1, 0]])

k2 = np.array([[1, 1, 1],
               [1, -8, 1],
               [1, 1, 1]])

L1 = cv2.filter2D(img, cv2.CV_32F, k1)
L2 = cv2.filter2D(img, cv2.CV_32F, k2)

# Sharpening
sharp = np.clip(img.astype(float) - L1, 0, 255)

plt.figure(figsize=(10, 7))

plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")

plt.subplot(2, 3, 2)
plt.imshow(gradient, cmap='gray')
plt.title("Gradient")

plt.subplot(2, 3, 3)
plt.imshow(np.abs(L1), cmap='gray')
plt.title("Laplacian 4-neighbour")

plt.subplot(2, 3, 4)
plt.imshow(np.abs(L2), cmap='gray')
plt.title("Laplacian 8-neighbour")

plt.subplot(2, 3, 5)
plt.imshow(sharp, cmap='gray')
plt.title("Sharpened Image")

plt.tight_layout()
plt.show()
```

```python
#13
import cv2
import matplotlib.pyplot as plt

path = input("Enter image path: ")

img = cv2.imread(path, 0)

if img is None:
    raise FileNotFoundError("Image not found")

t = int(input("Enter intensity threshold: "))

# Intensity thresholding
_, intensity = cv2.threshold(img, t, 255, cv2.THRESH_BINARY)

# Global thresholding using Otsu
T, global_img = cv2.threshold(
    img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

print("Global threshold:", T)

plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")

plt.subplot(2, 2, 2)
plt.imshow(intensity, cmap='gray')
plt.title("Intensity Thresholding")

plt.subplot(2, 2, 3)
plt.imshow(global_img, cmap='gray')
plt.title("Global Thresholding")

plt.subplot(2, 2, 4)
plt.hist(img.ravel(), bins=256, range=(0, 256))
plt.axvline(t, label="Intensity Threshold")
plt.axvline(T, label="Global Threshold")
plt.title("Image Histogram")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
#14
import cv2
import matplotlib.pyplot as plt

path = input("Enter image path: ")

img = cv2.imread(path)

if img is None:
    raise FileNotFoundError("Image not found")

h, w = img.shape[:2]

tx = int(input("Enter tx: "))
ty = int(input("Enter ty: "))

T = cv2.getRotationMatrix2D((0, 0), 0, 1)
T[0, 2] = tx
T[1, 2] = ty

translated = cv2.warpAffine(img, T, (w, h))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(translated, cv2.COLOR_BGR2RGB))
plt.title("Translation")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('image.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
rows, cols = img_rgb.shape[:2]

# Scaling factors
sx, sy = 1.5, 1.5

# 2x3 Scaling Matrix
M = np.float32([[sx, 0,  0],
                [0,  sy, 0]])

# Apply matrix (output size must be scaled to match new dimensions)
scaled = cv2.warpAffine(img_rgb, M, (int(cols * sx), int(rows * sy)))

plt.imshow(scaled)
plt.title(f"Matrix Scaled ({sx}x, {sy}x)")
plt.show()
```
